```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    "Student":["S1","S2","S3","S4","S5","S6","S7","S8","S9","S10"],
    "Study_hours":[2,3,4,5,6,7,8,9,4,5],
    "Attendence":[86,70,75,89,90,100,90,69,95,83],
    "Internal_Marks":[90,76,54,89,90,100,98,78,92,69],
    "Final_Marks":[90,77,89,73,88,100,79,86,96,70]
}
df=pd.DataFrame(data)
print(df)
```

```python
plt.hist(df["Final_Marks"], bins=5)
plt.xlabel("Final Marks")
plt.ylabel("Frequency")
plt.title("Histogram of Final Marks")
plt.show()
```

```python
plt.boxplot(df["Final_Marks"])
plt.xlabel("Final Marks")
plt.title("Boxplot of Final Marks")
plt.show()

```

```python
plt.boxplot([
    df['Study_hours'],
    df['Attendence'],
    df['Internal_Marks'],
    df['Final_Marks']
])

plt.xticks(
    [1,2,3,4],
    ["Study_hours","Attendence","Internal_Marks","Final_Marks"]
)
plt.ylabel("Marks")
plt.title("Boxplot of Marks")
plt.show()
```

```python
import seaborn as c
c.pairplot(
    df[
        [
            "Study_hours",
            "Attendence",
            "Internal_Marks",
            "Final_Marks"
        ]
    ]
)
plt.show()
```

```python
c.heatmap(df.corr(numeric_only=True),annot=True)
plt.title("Coorelation Heatmap")
plt.show()
```
